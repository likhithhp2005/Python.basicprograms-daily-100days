```python
import numpy as np

marks=np.array([[85,75,92],
               [74,88,90],
               [90,94,89],
               [65,70,72],
               [80,85,85]])
avg_student = np.mean(marks,axis=0)
print("average  marks in each subject :",avg_student)
print("highest marks ion each subject:",np.max(marks,axis=0))
print("lowest marks in each sub:",np.min(marks,axis=0))

topper = np.argmax(avg_student)
print("top-performing student(index):",topper)
print("topper's average marks:",avg_student[topper])
```

```python
import numpy as np
sales=np.array([
    [120,135,150,160,145,135],
    [80,95,100,110,105,115],
    [150,160,170 ,175,180,190]
])
total_sales=sales.sum(axis=1)
print("total sales:",total_sales)
avg_sales=sales.mean(axis=1)
print("Average sales:",avg_sales)

high_sales_month=sales.argmax(axis=1)
print("Highest sales:",high_sales_month)

best_product_index=total_sales.argmax()
print("best performing:",best_product_index)
```
