```python
students = {}

while True:
    print("\n1. Add student")
    print("2. Search student by ID")
    print("3. View all students")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        student_id = input("Enter student ID: ")
        student_name = input("Enter student name: ")
        students[student_id] = student_name
        print(f"Added student: {student_id} → {student_name}")

    elif choice == 2:
        student_id = input("Enter student ID to search: ")
        if student_id in students:
            print(f"Student found: {student_id} → {students[student_id]}")
        else:
            print("Student not found.")

    elif choice == 3:
        if students:
            print("Students (ID and Name):")
            for sid, name in students.items():
                print(f"{sid} → {name}")
        else:
            print("No students added yet.")

    elif choice == 4:
        print("Exiting...")
        break

    else:
        print("Invalid choice. Please try again.")
```

```python
products = {"ice cream":50, "book": 10, "pen":10, "glue stick": 25}
name = input("Enter the product name to get its price: ")
print("Price:", products.get(name, "Product not found"))
new = input("Enter a new product name to add: ")
new_price = input("Enter the price for the new product: ")
products.setdefault(new, int(new_price))
print("Updated Products:", products)
removed = products.popitem()
print("Removed Product:", removed)
print("All Products:", products.items())
```
